### Map
- Applies a given function to each item in an iterable and returns a map object (an iterator) containing the results.
- Syntax: map(function, iterable, ...)

```python
my_lst = [1,2,3,4,5]

def square(p_x):
    return(p_x*p_x)

result = list(map(square, my_lst)) #it is mapping the function to iterated to each item and convert it into list

print(result)
```

### filter
- Constructs an iterator from elements of an iterable for which a function returns true.
- Syntax: filter(function, iterable)

```python
my_lst = [1,2,3,4,5]

def square(p_x):
    if(p_x % 2 == 0):
        return(p_x*p_x)

result = list(map(square, my_lst)) 
print(result)
```

```python
# filter - will remove the none, none things.

my_lst = [1,2,3,4,5]

def square(p_x):
    if(p_x % 2 == 0):
        return(p_x*p_x)

result = list(filter(square, my_lst)) 
print(result)
```

### reduce
- Applies a function of two arguments cumulatively to the items of an iterable, from left to right, so as to reduce the iterable to a single value.
- Note: reduce() is part of the functools module and needs to be imported.
- Syntax: reduce(function, iterable, initializer)

```python
from functools import reduce

my_lst = [1,2,3,4,5]

def square(p_x, p_y):
    
        return(p_x+p_y)

result = reduce(square, my_lst)
print(result)
```

#OOP Concept
- It connect with the real-world.
- It provides the code-reuseability and better scalability.

Note - Even functions cannot handle reuseability (let's say we have 100 functions so we connot manage it) but bu the help OOP we can do it.

Class - A class is a blueprint or a template for creating objects. It defines a set of attributes (data) and methods (functions) that the objects created from it will possess. Think of it like a cookie cutter: the cutter itself is the class, defining the shape and characteristics of the cookies. 

Object - An object is an instance of a class. It is a concrete entity created from the class blueprint, possessing its own unique set of attribute values and able to execute the methods defined in its class. Using the cookie cutter analogy, the actual cookies you bake are the objects, each having the same shape but potentially different flavors or decorations.
Each object has its own set of attribute values.
Objects can call the methods defined in their class.

Attributes - These are variables defined within a class that represent the properties or characteristics of an object.

Methods - These are functions defined within a class that represent the actions or behaviors an object can perform.

### static method
 - like junk method, it'll not directly related to object or attributes but still it is useful.

 - we can use it to indirectly relate to other things.



```python
class employee:
    company_name = "XYZ"

    def __init__(self, emp_name, emp_dept):
        self.emp_name = emp_name
        self.emp_dept = emp_dept

    def changes(self, new_company):
        employee.company_name = new_company

    def info(self):
        print(f"Employee: {self.emp_name} works for {self.emp_dept} in {self.company_name} Organization")

    @staticmethod
    def addition(x,y):
        print(x+y)

emp1 = employee("Priya", "IT")  #emp1 is an Object
emp2 = employee("Priyanka", "Finance")

emp1.info()
emp2.info()

emp1.addition(4,5)
```

### Class Method
- It is originally used to interact with class attributes.

so we have special decorators to interact with the class attributes like:
@classmethod

```python
class employee:
    company_name = "XYZ"

    def __init__(self, emp_name, emp_dept):
        self.emp_name = emp_name
        self.emp_dept = emp_dept

    def changes(self, new_company):
        employee.company_name = new_company

    @classmethod
    def changesInClasses(cls, new_company):
        cls.company_name = new_company

    def info(self):
        print(f"Employee: {self.emp_name} works for {self.emp_dept} in {self.company_name} Organization")

    @staticmethod
    def addition(x,y):
        print(x+y)


emp1 = employee("Priya", "IT")  #emp1 is an Object
emp2 = employee("Priyanka", "Finance")

emp1.info()
emp2.info()
employee.addition(10,20)
emp1.changesInClasses("NewNew Deloitte")

print(emp1.company_name)
```

```python
class Employee:
    company_name = "XYZ"

    def __init__(self, emp_name, emp_dept):
        self.emp_name = emp_name
        self.emp_dept = emp_dept

    # Getter
    @property
    def info(self):
        return f"Employee: {self.emp_name} works for {self.emp_dept} in {self.company_name} Organization"

    # Setter
    @info.setter
    def info(self, new_details):
        self.emp_name = new_details[0]
        self.emp_dept = new_details[1]

    @classmethod
    def changesInClasses(cls, new_company):
        cls.company_name = new_company

    @staticmethod
    def addition(x, y):
        print(x+y)


emp1 = Employee("Priya", "IT")

# Update using setter
emp1.info = ["Sakshi", "HR"]

# Print using getter
print(emp1.info)
```

### Inheritance

# Multi-Threading
- Multithreading means doing multiple things at the same time in a program.

- It’s like having many small workers (threads) inside your program, each doing a task so that work finishes faster.

- In Python specifically

Python has something called the GIL (Global Interpreter Lock)
which allows only one thread to run Python code at a time.

So:

✔ Good for:

waiting tasks

downloading data

reading files

API calls

I/O operations

❌ Not good for:

heavy CPU tasks

mathematical calculations

processing large data

Because only one thread can run CPU-heavy code at once.

```python
import time
import random
from concurrent.futures import ThreadPoolExecutor

tables = ["orders", "products", "customers", "reviews", "cancels"]

def my_func(p_x):
   
    wait = random.randint(1, 10)
    time.sleep(wait)
    print(f"I'm {p_x}, I took {wait} seconds")

# instead of loops, we will use multi-thread becoz loops are slow
with ThreadPoolExecutor(max_workers=len(tables)) as executor:
    futures = executor.map(my_func, tables)
```

```python
import time
import random
from concurrent.futures import ThreadPoolExecutor

tables = ["orders", "products", "customers", "reviews", "cancels"]

def my_func(i):
   
    wait = 3
    time.sleep(wait)
    print(f"I'm {i}, I took {wait} seconds")

with ThreadPoolExecutor(max_workers=len(tables)) as executor:
    for i in tables:
        future = executor.submit(my_func, i)
```
